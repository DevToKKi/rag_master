```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
# 주어진 주제에 대해 짧은 설명을 요청하는 프롬프트 템플릿 정의
prompt = ChatPromptTemplate.from_template(
    "주제 {topic}에 대해 짧은 설명을 해주세요."
)
```

```python
# LM Studio 로컬 서버의 모델을 사용하여 채팅 모델 설정
model = ChatOpenAI(
    model="unsloth/gemma-4-e2b-it",                        # LM Studio에 로드된 모델명
    base_url="http://host.docker.internal:12345/v1",     # LM Studio 주소 + /v1
    api_key="lmstudio",                          # 임의 문자열
    temperature=0.7,
)
# 출력 파서를 문자열로 설정
output_parser = StrOutputParser()
```

```python
# 파이프라인 설정: 주제를 받아 프롬프트 생성 → 모델 응답 → 문자열 파싱
chain = (
    {"topic": RunnablePassthrough()}   # 입력받은 주제를 그대로 통과시킴
    | prompt                            # 프롬프트 템플릿 적용
    | model                             # 모델을 사용해 응답 생성
    | output_parser                     # 응답을 문자열로 파싱
)

# "더블딥" 주제로 설명 요청
chain.invoke("더블딥")
```
