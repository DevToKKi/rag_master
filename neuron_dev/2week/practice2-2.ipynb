```python
import os

# USER_AGENT 설정: 서버가 클라이언트 요청을 식별/추적하기 위함 (미설정 시 경고 발생)
os.environ["USER_AGENT"] = "MyApp/1.0 (Custom LangChain Application)"

from langchain_community.document_loaders import WebBaseLoader

# 단일 URL 초기화
loader = WebBaseLoader("https://docs.smith.langchain.com/")

# 다중 URL 초기화 (리스트)
loader_multiple_pages = WebBaseLoader(
    ["https://python.langchain.com/docs/introduction/",
     "https://langchain-ai.github.io/langgraph/"]
)
```

```python
# 단일 문서 로드
single_doc = loader.load()

# 메타데이터 확인 (source, title, language 등)
print(single_doc[0].metadata)
```

```python
# 다중 문서 로드
docs = loader_multiple_pages.load()

# 첫 번째 문서의 본문 확인
print(docs[0].page_content)
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.document_loaders import PyMuPDFLoader
from langchain_community.document_loaders import PDFPlumberLoader

# 실습 PDF 경로 (환경에 맞게 수정)
PDF_PATH = "2024 KB 부동산 보고서.pdf"
```

```python
%%time
loader = PyPDFLoader(PDF_PATH)
pages = loader.load_and_split()
print('청크의 수:', len(pages))
```

```python
pages[10]
```

```python
%%time
loader = PyMuPDFLoader(PDF_PATH)
pages = loader.load_and_split()
print('청크의 수:', len(pages))
```

```python
pages[10]
```

```python
%%time
loader = PDFPlumberLoader(PDF_PATH)
pages = loader.load_and_split()
print('청크의 수:', len(pages))
```

```python
pages[10]
```

```python
from langchain_community.document_loaders import CSVLoader
from langchain_community.document_loaders import UnstructuredCSVLoader

# 실습 CSV 경로 (환경에 맞게 수정)
CSV_PATH = "서울시_부동산_실거래가_정보.csv"
```

```python
%%time
loader = CSVLoader(CSV_PATH)
documents = loader.load()
print('청크의 수:', len(documents))
```

```python
documents[5]
```

```python
%%time
loader = UnstructuredCSVLoader(CSV_PATH, mode='elements')
documents = loader.load()
print('청크의 수:', len(documents))
```
